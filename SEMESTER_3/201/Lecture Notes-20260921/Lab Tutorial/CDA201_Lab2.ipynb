```python
import matplotlib.pyplot as plt

def plot_discrete_dist(dist_name,x_values,PMF,CDF):

    plt.figure(figsize=(10,4))
    # Plot PMF
    plt.subplot(1,2,1)
    plt.stem(x_values,PMF)
    plt.title(f"The plot of PMF ({dist_name})")
    plt.xlabel('X')
    plt.ylabel('P(X=x)')
    plt.grid(True)

    # Plot CDF
    plt.subplot(1,2,2)
    plt.step(x_values,CDF)
    plt.title(f"The plot of CDF ({dist_name})")
    plt.xlabel('X')
    plt.ylabel('P(X<=x)')
    plt.grid(True)

    plt.show()
```

```python
# The plot of Binomial(n,p)
import numpy as np
from math import comb

def Binomial_dist(n,p):
    x=np.arange(0,n+1)
    PMF=[comb(n,k)*(p**k)*((1-p)**(n-k)) for k in x]
    CDF=np.cumsum(PMF)
    return x,PMF, CDF

# Test Case
n=20
p=0.8

x_values,PMF,CDF=Binomial_dist(n,p)

plot_discrete_dist('Binomial Dist',x_values,PMF,CDF)
```

```python
from scipy.stats import poisson

lam = 5
x_p = np.arange(0, 50)

pmf_p = poisson.pmf(x_p, lam)
cdf_p = poisson.cdf(x_p, lam)
plot_discrete_dist('Poisson Dist',x_p,pmf_p,cdf_p)
```

```python
from math import exp, factorial
def Poisson_dist(n,lam):
    x_p=np.arange(0,n+1)
    PMF_p=[((float(lam)**k)*np.exp(-lam))/factorial(k) for k in x_p]
    CDF_p=np.cumsum(PMF_p)
    return x_p,PMF_p, CDF_p

# Use case
lam=20
n=50
x_p,pmf_p,cdf_p=Poisson_dist(n,lam)
plot_discrete_dist('Poisson Dist',x_p,pmf_p,cdf_p)
```

```python
import numpy as np
from scipy.stats import geom

upper_bound = 30
p=0.1
x_I = np.arange(1, upper_bound + 1)

PMF_I = geom.pmf(x_I, p)
CDF_I = geom.cdf(x_I, p)
plot_discrete_dist("Geometric Dist", x_I, PMF_I, CDF_I)
```

```python
def Geometric_dist(n,p):
    x_p=np.arange(1,n+1)
    PMF_p=[((1-p)**k)*p for k in x_p]
    CDF_p=np.cumsum(PMF_p)
    return x_p,PMF_p, CDF_p

# Use case
p=0.1
n=30
x_p,pmf_p,cdf_p=Geometric_dist(n,p)
plot_discrete_dist('Geometric Dist',x_p,pmf_p,cdf_p)
```

```python
def Binomial_dist_exp(n,p):
    x=np.arange(0,n+1)
    PMF=[comb(n,k)*(p**k)*((1-p)**(n-k)) for k in x]
    exp_x=np.sum([k*PMF[k] for k in x])
    exp_xx=np.sum([(k**2)*PMF[k] for k in x])
    Var_x=exp_xx-exp_x**2
    return exp_x,Var_x

n=20
p=0.5
mean,var=Binomial_dist_exp(n,p)
print(f"The mean and variance of the given distribution are {mean} and {var}")
```
