```python
import matplotlib.pyplot as plt

def plot_conti_dist(dist_name,x_values,PDF,CDF):

    plt.figure(figsize=(10,4))
    # Plot PDF
    plt.subplot(1,2,1)
    plt.plot(x_values,PDF)
    plt.title(f"The plot of PDF ({dist_name})")
    plt.xlabel('X')
    plt.ylabel('f(X=x)')
    plt.grid(True)

    # Plot CDF
    plt.subplot(1,2,2)
    plt.plot(x_values,CDF)
    plt.title(f"The plot of CDF ({dist_name})")
    plt.xlabel('X')
    plt.ylabel('P(X<=x)')
    plt.grid(True)

    plt.show()
```

```python
from scipy.stats import norm
import numpy as np

mu=3
sigma=3
x_val=np.linspace(mu-5*sigma,mu+5*sigma,100)
# x_val=np.linspace(0,10,12)
PDF_n=norm.pdf(x_val,loc=mu,scale=sigma)
CDF_n=norm.cdf(x_val,loc=mu,scale=sigma)
plot_conti_dist("Normal",x_val,PDF_n,CDF_n)
```

```python
from scipy.stats import norm
from scipy.special import erf
import numpy as np

def normal_dist(mu,sigma):
    x_val=np.linspace(mu-5*sigma,mu+5*sigma,100)
    PDF=(1/np.sqrt(2*np.pi))*np.exp((-(x_val-mu)**2)/(2*(sigma**2)))

    z=(x_val-mu)/(np.sqrt(2)*sigma)
    CDF=0.5+0.5*erf(z)
    return x_val,PDF,CDF
```

```python
mu=3
sigma=3
x_values,PDF,CDF=normal_dist(mu,sigma)
plot_conti_dist("Normal",x_values,PDF,CDF)
```

```python
def expo_dist(lamda):
    x_val=np.linspace(0,10,100)
    PDF=lamda*np.exp(-lamda*x_val)
    CDF=1-np.exp(-lamda*x_val)
    return x_val,PDF,CDF

lamda=0.5
x_values,PDF,CDF=expo_dist(lamda)
plot_conti_dist("Exponential_distribution",x_values,PDF,CDF)
```

```python
from scipy.stats import expon
lamda=0.5
x_values=np.linspace(0,10,100)
PDF=expon.pdf(x_values,scale=1/lamda)
CDF=expon.cdf(x_values,scale=1/lamda)

plot_conti_dist("Exponential_distribution",x_values,PDF,CDF)
```

```python
def uniform_dist(a, b):
    x_val = np.linspace(a - (b - a) / 4, b + (b - a) / 4, 100)

    pdf = np.zeros_like(x_val)
    cdf = np.zeros_like(x_val)

    # Calculate PDF
    pdf_value = 1 / (b - a) if b > a else 0
    pdf[(x_val >= a) & (x_val <= b)] = pdf_value

    # Calculate CDF
    cdf[x_val >= a] = (x_val[x_val >= a] - a) / (b - a)
    cdf[x_val > b] = 1

    return x_val, pdf, cdf

# Case for Uniform Distribution
a_uniform = 0
b_uniform = 10
x_uniform, pdf_uniform, cdf_uniform = uniform_dist(a_uniform, b_uniform)
plot_conti_dist(f"Uniform Dist [{a_uniform}, {b_uniform}]", x_uniform, pdf_uniform, cdf_uniform)
```
