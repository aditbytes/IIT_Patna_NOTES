```python
# Rolling of two dices
import numpy as np

while True:
  input_=input("Roll the dices (y/n)")
  if input_=="n":
    print("Thanks for playing")
    break
  elif input_=="y":
    a=np.random.randint(1,7)
    b=np.random.randint(1,7)
    print((a,b))
  else:
    print("Invalid choice!")
```

```python
# Guess the number
import random

number = random.randint(0, 100)

while True:

    guess = int(input("Enter any number between 0 to 100: "))

    if guess < 0 or guess > 100:
        print("Enter the valid number")
    elif guess < number:
        print("The number you have entered is too low")
    elif guess > number:
        print("The number you have entered is too high")
    else:
        print("Congratulations! You guessed the correct number {number}")
        break
```

```python
choices = ["r", "p", "s"]

while True:
    computer = random.choice(choices)
    user = input("Enter your choice (r,p,s):").lower()

    print("You have chosen", user)
    print("Computer has chosen", computer)

    if user == computer:
        print("Game is tie")

    elif (user == "r" and computer == "s") or \
         (user == "p" and computer == "r") or \
         (user == "s" and computer == "p"):
        print("You won!")

    else:
        print("You lost")

    play_again = input("Do you want to play again? (y/n):").lower()

    if play_again == "n":
         print("Thank you!")
         break
```

```python
import re

password = input("Enter your password: ")
score = 0

if len(password)>=8:
    score += 1
if re.search(r"[A-Z]", password):
    score += 1
if re.search(r"[a-z]", password):
    score += 1
if re.search(r"[0-9]", password):
    score += 1
if re.search(r"[@#$%&!*^()_+\-=\[\]{};:'\",.<>?/\\|`~]", password):
    score += 1

print("Points scored:", score)

if score == 5:
    print("Password Strength: Strong")
elif score == 4:
    print("Password Strength: Moderate")
elif score == 3:
    print("Password Strength: Less Protected")
else:
    print("Password Strength: Very Poor")
```
